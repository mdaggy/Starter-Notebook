## 3. Quick look at the data (2-3 real numbers)
The dataset contains 3,000 housing records. The average median income is 3.81, while the average median house value is 205,846.28. Median income has a correlation of 0.67 with median house value, which is stronger than the correlations observed for most of the other variables. These initial results suggest that the relationship between income and house value is worth investigating further.

```python
import pandas as pd

file_path = "/content/sample_data/california_housing_test.csv"

df = pd.read_csv(file_path)

print("Number of records:", len(df))

print("Average median income:", round(df["median_income"].mean(), 2))

print("Average median house value:", round(df["median_house_value"].mean(), 2))

print("Average total rooms:", round(df["total_rooms"].mean(), 2))

print("\nCorrelation with median house value:")
print(
    df.corr(numeric_only=True)["median_house_value"]
    .sort_values(ascending=False)
)
```
