## 3. Success metric

I will use **Mean Absolute Error (MAE)** as the success metric. MAE measures the average absolute difference between the estimated and observed median house values, so lower is better. I will consider a model useful if its MAE is lower than a simple baseline that always uses the average house value as the estimate.


```python
import pandas as pd
from sklearn.metrics import mean_absolute_error

file_path = "/content/sample_data/california_housing_test.csv"

df = pd.read_csv(file_path)

y = df["median_house_value"]

baseline_prediction = y.mean()

baseline_mae = mean_absolute_error(
    y,
    [baseline_prediction] * len(y)
)

print("Baseline prediction:", round(baseline_prediction, 2))
print("Baseline MAE:", round(baseline_mae, 2))
print("Lower MAE is better.")
```
